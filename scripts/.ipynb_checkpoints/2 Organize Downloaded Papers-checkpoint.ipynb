```python
import os
import pandas as pd
```

## Data

```python
input_dir = '/Users/jimuelcelestejr/Documents/codebook/transparency_simulations/data/2_survey_papers'
input_metadata = '/Users/jimuelcelestejr/Documents/codebook/transparency_simulations/data/2_survey_papers/Transparency Simulations.csv'
```

```python
df = pd.read_csv(input_metadata)
df.columns
```

```python
df['filename'] = df['File Attachments'].apply(lambda x: os.path.basename(x))
df['filename']
```

'DOI' and 'File Attachments' will be used to map the metadata (from step 1) to the downloaded files. There should be 100 files.

```python
df.shape
```

```python
df = df[['DOI', 'filename']]
df.head()
```

## Mapping

```python
metadata = pd.read_csv("/Users/jimuelcelestejr/Documents/codebook/transparency_simulations/output/metadata.csv")
metadata.columns
```

```python
df['DOI'] = df['DOI'].apply(lambda doi: doi.lower())
metadata['DOI'] = metadata['DOI'].apply(lambda doi: doi.lower())
```

```python
set(metadata['DOI'].values) - set(df['DOI'].values)
```

```python
set(df['DOI'].values) - set(metadata['DOI'].values)
```

There is 1:1 mapping.

```python
output = pd.merge(metadata, df, on='DOI', how='inner')
output.shape
```

```python
len(output['DOI'].unique())
```

There are 100 unique DOIs.

## File checking

A checkpoint to check if the papers exist. 

```python
metadata_files = output['filename'].values
metadata_files
```

```python
for _, _, files in os.walk(input_dir):
    for file in files:
        if file not in metadata_files:
            print(file)
```

These are metadata files. The list is complete.

## Save output

```python
output_file = os.path.join(input_dir, "metadata.csv")
output.to_csv(output_file, index=False)
```
